```python
import pandas as pd

file_name = "Day7_Student_dataset.csv"
student_data = pd.read_csv("Day7_Student_Dataset.csv")
print("Dataset loaded successfully.")

print("========== COMPLETE DATASET ==========")
display(student_data)

print("========== FIRST 10 ROWS ==========")
display(student_data.head(10))

print("========== LAST 10 ROWS ==========")
display(student_data.tail(10))

print("========== RANDOM SAMPLE ==========")
display(student_data.sample(5))

print("========== DATASET SHAPE ==========")
number_of_rows = student_data.shape[0]
number_of_columns = student_data.shape[1]
print(f"Number of rows: {number_of_rows}")
print(f"Number of columns: {number_of_columns}")
print(f"Dataset shape: {student_data.shape}")

print("========== COLUMN NAMES ==========")
column_names = student_data.columns.tolist()
for column_name in column_names:
  print(column_name)

print("========== DATASET INDEX ==========")
print(student_data.index)
print("\nIndex values:")
print(student_data.index.tolist())

print("========== DATA TYPES ==========")
print(student_data.dtypes)

print("========== DATASET INFORMATION ==========")
student_data.info()

print("========== DESCRIPTIVE STATISTICS ==========")
display(student_data.describe())

print("========== ALL COLUMN STATISTICS ==========")
display(student_data.describe(include="all"))

print("========== MISSING VALUES BY COLUMN ==========")
missing_values = student_data.isnull().sum()
print(missing_values)

print("========== TOTAL MISSING VALUES ==========")
total_missing_values = student_data.isnull().sum().sum()
print(f"Total missing values: {total_missing_values}")

print("========== ROWS WITH MISSING VALUES ==========")
missing_rows = student_data[
    student_data.isnull().any(axis=1)
]
display(missing_rows)

print("========== UNIQUE VALUES ==========")
unique_values = student_data.nunique()
print(unique_values)

print("========== STUDENT BY COURSE ==========")
course_counts = student_data["Course"].value_counts()
print(course_counts)

print("========== STUDENTS BY CITY ==========")
city_counts = student_data["City"].value_counts()
print(city_counts)

print("========== AGE ANALYSIS ==========")
average_age = student_data["Age"].mean()
median_age = student_data["Age"].median()
minimum_age = student_data["Age"].min()
maximum_age = student_data["Age"].max()
print(f"Average age: {average_age:.2f}")
print(f"Median age: {median_age:.2f}")
print(f"Minimum age: {minimum_age}")
print(f"Maximum age: {maximum_age}")

print("========== MARKS ANALYSIS ==========")
average_marks = student_data["Marks"].mean()
median_marks = student_data["Marks"].median()
minimum_marks = student_data["Marks"].min()
maximum_marks = student_data["Marks"].max()
print(f"Average marks: {average_marks:.2f}")
print(f"Median marks: {median_marks:.2f}")
print(f"Minimum marks: {minimum_marks:.2f}")
print(f"Maximum marks: {maximum_marks:.2f}")

print("========== ATTENDANCE ANALYSIS ==========")
average_attendance = (
    student_data["Attendance_Percentage"].mean()
)
median_attendance = (
    student_data["Attendance_Percentage"].median()
)
minimum_attendance = (
    student_data["Attendance_Percentage"].min()
)
maximum_attendance = (
    student_data["Attendance_Percentage"].max()
)
print(f"Average attendance: {average_attendance:.2f}%")
print(f"Median attendance: {median_attendance:.2f}%")
print(f"Minimum attendance: {minimum_attendance:.2f}%")
print(f"Maximum attendance: {maximum_attendance:.2f}%")

print("========== HIGHEST-SCORING STUDENT ==========")
highest_marks_index = student_data["Marks"].idxmax()
highest_marks_student = student_data.loc[highest_marks_index]
display(highest_marks_student)

print("========== LOWEST-SCORING STUDENT ==========")
lowest_marks_index = student_data["Marks"].idxmin()
lowest_marks_student = student_data.loc[lowest_marks_index]
display(lowest_marks_student)

print("========== MOST POPULAR COURSE ==========")
most_popular_course = (
    student_data["Course"].value_counts().idxmax()
)
course_student_count = (
    student_data["Course"].value_counts().max()
)
print(f"Most popular course: {most_popular_course}")
print(f"Number of students: {course_student_count}")

print("========== MOST REPRESENTED CITY ==========")
most_common_city = (
    student_data["City"].value_counts().idxmax()
)
city_student_count = (
    student_data["City"].value_counts().max()
)
print(f"Most represented city: {most_common_city}")
print(f"Number of students: {city_student_count}")

print("=" * 60)
print("                     FINAL DATASET SUMMARY")
print("=" * 60)
print(f"Total students: {number_of_rows}")
print(f"Total columns: {number_of_columns}")
print(f"Average age: {average_age:.2f}")
print(f"Average marks: {average_marks:.2f}")
print(f"Median marks: {median_marks:.2f}")
print(f"Highest marks: {maximum_marks:.2f}")
print(f"Lowest marks: {minimum_marks:.2f}")
print(f"Average attendance: {average_attendance:.2f}%")
print(f"Total missing values: {total_missing_values}")
print(f"Most popular course: {most_popular_course}")
print(f"Most represented city: {most_common_city}")
print("=" * 60)
```
